```python
import os
import shutil

folder_path=r"C:\\Users\\SAVITRI\\Downloads\\Automatic_File_Organizer_Sample_Files (1)\\Automatic_File_Organizer"
print("File Organizer Started !")
print(os.listdir(folder_path))
```

```python
for file in os.listdir(folder_path):
    if file.endswith(".pdf"):
        print(file, " - PDF File")

    elif file.endswith(".xlsx"):
        print(file, " - Excel File")

    elif file.endswith((".jpg", ".png")):
        print(file, " - Image File")

    elif file.endswith(".docx"):
        print(file, " - Word File")

    elif file.endswith(".txt"):
        print(file, " - Text File")
```

```python
folders = ["PDF", "Excel", "Images", "Word", "Text"]

for folder in folders:
    path = os.path.join(folder_path, folder)

    if not os.path.exists(path):
        os.mkdir(path)

    print(folder, "Folder Created Successfully!")
```

```python
for file in os.listdir(folder_path):

    source = os.path.join(folder_path, file)

    if os.path.isfile(source):

        if file.endswith(".pdf"):
            destination = os.path.join(folder_path, "PDF", file)

        elif file.endswith(".xlsx"):
            destination = os.path.join(folder_path, "Excel", file)

        elif file.endswith((".jpg", ".png")):
            destination = os.path.join(folder_path, "Images", file)

        elif file.endswith(".docx"):
            destination = os.path.join(folder_path, "Word", file)

        elif file.endswith(".txt"):
            destination = os.path.join(folder_path, "Text", file)

        else:
            continue

        shutil.move(source, destination)

        print(file, "Moved Successfully!")
```

```python
for folder in folders:

    path = os.path.join(folder_path, folder)

    print("\n", folder, "Folder:")

    files = os.listdir(path)

    for file in files:
        print("  ", file)
```

```python
total_files = 0

for folder in folders:

    path = os.path.join(folder_path, folder)

    count = len(os.listdir(path))

    total_files = total_files + count

    print(folder, ":", count, "Files")

print("----------------------")
print("Total Organized Files:", total_files)
print("File Organization Completed!")
```

```python
import pandas as pd

report = []

for folder in folders:

    path = os.path.join(folder_path, folder)

    for file in os.listdir(path):

        report.append({
            "File Name": file,
            "Category": folder
        })

df = pd.DataFrame(report)

df.to_csv(
    os.path.join(folder_path, "File_Organizer_Report.csv"),
    index=False
)

print("Report Saved Successfully!")
display(df)
```
